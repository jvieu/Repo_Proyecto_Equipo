```python
import pandas as pd
import altair as alt
archivo = "charts_latam_carol_6m.csv"

df = pd.read_csv(archivo)

df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df[["pais", "fecha", "mes_muestra", "rank", "artist_names", "track_name", "streams"]].head()
```

```python
artist_cols = [f"artist_{str(i).zfill(2)}" for i in range(1, 11)]

artist_cols
```

```python
df_long = df.melt(
    id_vars=["pais", "fecha", "mes_muestra", "rank", "track_name", "streams"],
    value_vars=artist_cols,
    var_name="artist_posicion",
    value_name="artista"
)

df_long.head()
```

```python
df_long["artista"] = df_long["artista"].astype(str).str.strip()

df_long = df_long[
    (df_long["artista"].notna()) &
    (df_long["artista"] != "") &
    (df_long["artista"] != "NA") &
    (df_long["artista"].str.lower() != "nan")
]

df_long.head()
```

```python
base_vis = (
    df_long
    .groupby(["artista", "pais"], as_index=False)
    .agg(
        apariciones=("track_name", "count"),
        meses_presente=("mes_muestra", "nunique"),
        streams_total=("streams", "sum")
    )
)

base_vis.head()
```

```python
top_artistas = (
    base_vis
    .groupby("artista", as_index=False)
    .agg(
        apariciones_totales=("apariciones", "sum"),
        paises_distintos=("pais", "nunique"),
        meses_distintos=("meses_presente", "sum"),
        streams_total=("streams_total", "sum")
    )
    .sort_values("apariciones_totales", ascending=False)
    .head(15)
)

top_artistas
```

```python
base_vis_top = base_vis[base_vis["artista"].isin(top_artistas["artista"])]

base_vis_top.head()
```

```python
base_vis_top.to_csv(
    "base_visualizacion_coincidencias_artistas.csv",
    index=False,
    encoding="utf-8"
)
```

```python
chart = alt.Chart(base_vis_top).mark_circle().encode(
    x=alt.X(
        "pais:N",
        title="País"
    ),
    y=alt.Y(
        "artista:N",
        title="Artista",
        sort=alt.EncodingSortField(
            field="apariciones",
            op="sum",
            order="descending"
        )
    ),
    size=alt.Size(
        "apariciones:Q",
        title="Apariciones",
        scale=alt.Scale(range=[80, 900])
    ),
    color=alt.Color(
        "apariciones:Q",
        title="Apariciones"
    ),
    tooltip=[
        alt.Tooltip("artista:N", title="Artista"),
        alt.Tooltip("pais:N", title="País"),
        alt.Tooltip("apariciones:Q", title="Apariciones"),
        alt.Tooltip("meses_presente:Q", title="Meses presente"),
        alt.Tooltip("streams_total:Q", title="Streams acumulados", format=",")
    ]
).properties(
    title="Los artistas que cruzan fronteras en los charts latinoamericanos",
    width=650,
    height=450
)

chart
```
